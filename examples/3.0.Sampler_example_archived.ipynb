## Принципы работы с sampler'ом

Главной задачей sampler'а является хранение и предобработка данных. Предобработка реализована в конструкторе. Реализация предобработки зависит от конкретного датасета и того формата данных, которые прислали контрагенты. 

Тестовый датасет должен лежать в аттрибуте `test`, валидационный - в аттрибуте `val`, а обучающий - в `train`. Датасет состоит из трех частей: `X` - dataFrame с вопросами, ответами, контекстом и т.п.; `y` - dataFrame, необходимый для подсчета метрик; `traces` - dataFrame с распаршенными трейсами формата aef-tracing (опционально).

Существует 2 способа работы с семплером:
1. Можно написать предобработку данных вне семплера и запихнуть данные в дефолтный семплер (***Нецелевой***);
2. Создается класс, и в нем описывается логика предобработки данных (***Целевой***).

Для каждого формата датасетов определена (***пока не определена***) строгая структура для корректной работы с библиотекой `llm_val`.

**MMLU**:
- ***X***:
    1. *system* - Системный промпт. `str`;
    2. *question* - Вопрос. `str`;
    3. *context* - Список с контекстами. `List[str]`;
    4. *thema* - Тематика вопроса. `str`;
    5. *answer_options* - Словарь с вариантами ответов. `Dict[str, str]`;
    6. *true_answer* - Правильный ответ. Одна латинская буква. `str`;
    7. *raw_prompt* - Весь запрос, который целиком будет передаваться в модель. `str`;
    8. *model_answer* - Распаршенный ответ модели. Одна латинская буква. `str`;
    9. *few_shot* - История диалога с fewshot-подсказками для модели в целевом формате GigaChat API (пример в ячейке ниже). `Tuple[Dict[str,str]]`.
- ***y***:
    1. *metric_value* - Посчитанная метрика на конкретном вопросе. 1, если `true_answer == model_answer`, иначе 0. `Union[float, int]`.

**RuBQ**:
- ***X***:
    1. *system* - Системный промпт. `str`;
    2. *question* - Вопрос. `str`;
    3. *context* - Список с контекстами. `List[str]`;
    4. *thema* - Тематика вопроса. `str`;
    5. *true_answer* - Список правильных ответов. `List[str]`.
    6. *raw_prompt* - Весь запрос, который целиком будет передаваться в модель. `str`;
    7. *model_answer* - Распаршенный ответ модели. `str`;
    8. *few_shot* - История диалога с fewshot-подсказками для модели в целевом формате GigaChat API (пример в ячейке ниже). `Tuple[Dict[str,str]]`.
- ***y***:
    1. *metric_value* - Посчитанная метрика на конкретном вопросе. Например, если нас интересует только совпадение, то 1, если `true_answer` в том или ином виде содержится в `answer`, и 0 иначе. `Union[float, int]`.

**LongAnswer**:
- ***X***:
    1. *system* - Системный промпт. `str`;
    2. *question* - Вопрос. `str`;
    3. *context* - Список с контекстами. `List[str]`;
    4. *thema* - Тематика вопроса. `str`;
    5. *true_answer* - Список релевантных пассажей. `List[str]`;
    6. *raw_prompt* - Весь запрос, который целиком будет передаваться в модель. `str`;
    7. *model_answer* - Распаршенный ответ модели. `str`;
    8. *<название_критерия>* - n-ое кол-во столбцов с оценкой асессоров по критериям. В каждом столбце находятся словари, где ключ - id асессора, а значение - оценка по критерию. `Dict[str, float]`;
    9. *few_shot* - История диалога с fewshot-подсказками для модели в целевом формате GigaChat API (пример в ячейке ниже). `Tuple[Dict[str,str]]`.
- ***y***:
    1. *<название_критерия>* - Посчитанная метрика на конкретном вопросе (В случае LongAnswer, например, по алгоритму Дэвида-Скина). `Union[float, int]`.

**RAG**:
- ***X***
    1. *system* - Системный промпт. `str`;
    2. *question* - Вопрос. `str`;
    3. *context* - Список с контекстами. `List[str]`;
    4. *thema* - Тематика вопроса. `str`;
    5. *true_answer* - Список релевантных пассажей. `List[str]`;
    6. *model_answer* - Распаршенный ответ модели. `str`;
    7. *passage_accuracy* - Словарь с оценками релевантностей пассажей. `Dict[Union[str, int], List[int]]`;
    8. *<название_критерия>* - n-ое кол-во столбцов с оценкой асессоров по критериям. В каждом столбце находятся словари, где ключ - id асессора, а значение - оценка по критерию. `Dict[str, float]`.

- ***y***
    1. *sim_<название_расстояния>* - Расстояния от вопроса до пассажей. Размер списка n, где n - количество пассажей. `List[float]`;
    2. *criterion_<названия_критерия>* - Посчитанная метрика на конкретном вопросе(аггрегация метков асессоров в разрезе вопроса). `Union[float, int]`.

**Chat:**
- ***X***:
    1. *dialogue_id* - id диалога. `int`;
    2. *content* - Текст сообщения. `str`;
    3. *embedding* - Опционально: эмбеддинги сообщений. Для получения эмбеддингов можно использовать любую модель, которая отображает предложения в вектор. `List[np.array]`;
    4. *pos_id* - Индекс сообщения в диалоге. `int`;
- ***y***:
    1. *dialogue_id* - id Диалога. `int`;
    2. *role* - Роль автора сообщения, может быть 'user' и 'assistant'. `str`;
    3. *pos_id* - Индекс сообщения в диалоге. `int`;
    
**Rewrite**:
- ***X***
    1. *system* - Системный промпт. `str`;
    2. *question* - Текст для переписывания. `str`;
    3. *context* - Список с контекстами. `List[str]`;
    4. *thema* - Тематика вопроса. `str`;
    5. *true_answer* - Правильный результат переписывания исходного текста. `str`;
    6. *model_answer* - Результат переписывания при решении моделью прямой задачи. `str`;
    7. *reverse_problem_answer* - Результат переписывания при решении моделью обратной задачи, `str`.
    
**NER**:
- ***Атрибуты***:
    1. *entity_types* - Список типов (классов) сущностей для извлечения. `List[str]`;
- ***X***:
    1. *id* - Идентификатор примера. `Union[str, int]`;
    2. *text* - Текст для извлечения сущностей. `str`;
    3. *extracted* - Извлеченные сущности. `Dict[str, List[str]]`;
    4. *target* - Целевые сущности. `Dict[str, List[str]]`;
    5. *parser_error* - Флаг ошибки парсинга генерации (предсказания) модели в структурированный формат. `bool`;
    6. *flat_extracted_entities* - Список всех извлеченных сущностей в плоском виде. `List[str]`;
    7. *flat_target_entities* - Список всех целевых сущностей в плоском виде. `List[str]`;
    8. *extracted_all_types* - Извлеченные сущности при извлечении всех типов за раз. `Dict[str, List[str]]`;
    9. *extracted_by_type_merged* - Объединенные извлеченные сущности при извлечении каждого типа отдельно. `Dict[str, List[str]]`;
    10. *extracted_description_detalization* - Извлеченные сущности при детализации описаний классов. `Dict[str, List[str]]`;
    11. *extracted_examples_addition* - Извлеченные сущности при добавлении дополнительных примеров. `Dict[str, List[str]]`;
    12. *extracted_description_detalization_and_examples_addition* - Извлеченные сущности при детализации описаний классов и добавлении дополнительных примеров. `Dict[str, List[str]]`;
- ***y***:
    1. *extracted* - Извлеченные сущности. `Dict[str, List[str]]`;
    2. *target* - Целевые сущности. `Dict[str, List[str]]`.

- ***Комментарии***:
    1. Формат извлеченных и целевых сущностей может адаптироваться под конкретное применение (кроме flat_* полей), при этом необходимо реализовать соответствующие новому формату функции расчета метрик.
 
**Memory**:
- ***X***:
    1. *request_id* - id диалога пользователь-агент. `str`;
    2. *order_idx* - порядковый номер сообщения для хронологической сортировки. `int`;
    3. *question* - Запрос пользователя. `str`;
    4. *model_answer* - Ответ агента. `str`;
    5. *reasoning* - Внутренние рассуждения агента. `str`;
    6. *history_summarization* - Суммаризированная история диалога. `str`;
- ***y***:
    1. *<название_критерия>* - Посчитанная метрика на конкретном вопросе. 
- ***Комментарии***:
    1. Для колонок *y* и *history_summarization* обязатально наличие данных в хронологически последней *order_idx* строке диалога *request_id* и опционально во всех остальных.

**Трейсы** - опционально можно добавить в семплер для любого формата датасета.
- ***X***: исходя из описания выше под соответствующую задачу;
- ***y***: исходя из описания выше под соответствующую задачу;
- ***traces***:
    1. *traceid* - id трейса, соответствующее одному диалогу/запросу;
    2. *spanid* - id спана, соответствующее одной ноде/инструменту агента;
    3. *name* - название ноды/инструмента агента;
    4. *starttimeunixnano* - время начала работы ноды агента в unix-формате;
    5. *endtimeunixnano* - время окончания работы ноды агента в unix-формате.

## BrainStormSamplerUpgraded

`BrainStormSamplerUpgraded` является примером реализации sampler'а с предобработкой данных. 

Применим в случае, если данные имеют следующие обязательные колонки: 
- `system_prompt` - Строка, содержащая системный промпт. `str`;
- `task` - Строка запроса для генерации идей. `str`;
- `ideas` - Строка, содержащая необработанный ответ модели на запрос. `str`;
- `theme` - Тематика запроса. `str`;
- `per_idea_score` - Список дискретных асессорских оценок для каждой идеи в рамках одного запроса. `list[int]`.

На выходе ожидается датафрейм со следующим минимальным набором колонок:
 - `question` - содержимое `task` из датасета. `str`;
 - `model_answer` - содержимое `ideas` из датасета. `str`;
 - `thema` - содержимое `theme` из датасета. `str`;
 - `system` - содержимое `system_prompt` из датасета. `str`;
 - `per_idea_score` - содержимое `per_idea_score` из датасета. `list[int]`;
 - `total_score` - агрегированная взвешенная оценка на базе `per_idea_score`. `float`;
 - `context` - поле-наследник LongAnswer формата датасета. `str`;
 - `true_answer` - поле-наследник LongAnswer формата датасета. `str`;
 - `splitted_ideas` - разделенные за явно заданному паттерну идеи из `model_answer`. `list[str]`;
 - `n_ideas` - кол-во идей в `splitted_ideas` после разделения. `int`.
  
  Стоит отметить, что данные оформлены в том же формате, что и в BrainStormSampler, что важно, поскольку в предобработку вшита регулярка для разделения идей между собой. Также данный sampler имеет `train`-поле, которое предназначено для того, чтобы на нем калибровать пороги значений мер близости для их биннинга в дискретные значения.
